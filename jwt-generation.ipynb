```python
import secrets
import base64

# Générer une clé de 64 octets
secret_key = secrets.token_hex(32)  # 32 bytes = 64 caractères hex
print(f"Clé hexadécimale: {secret_key}")

# Ou en base64
secret_key_base64 = base64.urlsafe_b64encode(secrets.token_bytes(64)).decode()
print(f"Clé base64: {secret_key_base64}")
```

```python
import os
import base64

def generate_jwt_secret(length: int = 64) -> str:
    """
    Génère une clé secrète JWT en Base64.
    length = taille en octets avant encodage (64 = 512 bits).
    """
    secret_bytes = os.urandom(length)  # Génère des octets aléatoires sécurisés
    secret_b64 = base64.b64encode(secret_bytes).decode("utf-8")  # Encode en Base64 standard
    return secret_b64

if __name__ == "__main__":
    secret = generate_jwt_secret()
    print("Clé JWT (Base64 64 octets) :")
    print(secret)
```
